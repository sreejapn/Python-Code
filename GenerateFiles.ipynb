```python
import csv
import shutil
import os

accounts_csv = "accounts.csv"
ranges_csv = "ranges.csv"
template_pdf = "sample.pdf"
output_dir = "eStatements"

# Create output directory
os.makedirs(output_dir, exist_ok=True)

# Read account numbers
with open(accounts_csv, newline="") as acc_file:
    acc_reader = csv.DictReader(acc_file)
    accounts = [row["AccNo"] for row in acc_reader]

# Read date ranges
with open(ranges_csv, newline="") as range_file:
    range_reader = csv.DictReader(range_file)
    ranges = list(range_reader)

count = 0

# Generate PDFs
for acc_no in accounts:
    for r in ranges:
        from_date = r["FromDate"]
        to_date = r["Todate"]
        customer_type = r["CustomerType"]
        account_type = r["AccountType"]

        filename = (
            f"{acc_no}_{from_date}_{to_date}_"
            f"{customer_type}_{account_type}.pdf"
        )

        dest_path = os.path.join(output_dir, filename)

        # Exact replica of sample.pdf
        shutil.copyfile(template_pdf, dest_path)

        count += 1

print(f"✅ Total PDFs created: {count}")
```
